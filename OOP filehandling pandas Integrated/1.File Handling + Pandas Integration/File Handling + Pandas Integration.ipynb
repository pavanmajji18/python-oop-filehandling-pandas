```python
import os
import pandas as pd

# 1. Create and write raw sales.txt
raw_data = """order_id,product,quantity,price,city
101,Laptop,1,65000,Bangalore
102,Mouse,2,1200,Delhi
103,Keyboard,1,3500,Mumbai
104,Monitor,2,14000,Chennai
105,Headphones,3,2500,Bangalore
106,Chair,1,8500,Hyderabad
107,Desk,1,12000,Pune
108,Mousepad,4,450,Delhi
109,Webcam,2,4200,Mumbai
110,Microphone,1,6800,Bangalore
111,Laptop,2,65000,Delhi
112,Mouse,5,1200,Hyderabad
113,Keyboard,2,3500,Chennai
114,Monitor,1,14000,Pune
115,Headphones,1,2500,Delhi
116,Chair,2,8500,Bangalore
117,Desk,1,12000,Mumbai
118,Mousepad,3,450,Chennai
119,Webcam,1,4200,Hyderabad
120,Microphone,2,6800,Pune
121,Laptop,1,65000,Mumbai
122,Mouse,3,1200,Bangalore
123,Keyboard,1,3500,Delhi
124,Monitor,3,14000,Hyderabad
125,Headphones,2,2500,Chennai
126,Chair,1,8500,Pune
127,Desk,2,12000,Bangalore
128,Mousepad,5,450,Mumbai
129,Webcam,2,4200,Delhi
130,Microphone,1,6800,Chennai"""

with open("sales.txt", "w", encoding="utf-8") as f:
    f.write(raw_data)
```

```python
# 2. Read using standard Python file handling
print("--- Standard File Read (First 3 Lines) ---")
with open("sales.txt", "r", encoding="utf-8") as f:
    for _ in range(3):
        print(f.readline().strip())
```

```python
# 3. Load using Pandas & Convert into DataFrame
df = pd.read_csv("sales.txt")
```

```python
# 4. Calculations
df["revenue"] = df["quantity"] * df["price"]

total_revenue = df["revenue"].sum()
avg_order_value = df["revenue"].mean()
highest_order = df.loc[df["revenue"].idxmax()]
lowest_order = df.loc[df["revenue"].idxmin()]
product_revenue = df.groupby("product")["revenue"].sum()
city_revenue = df.groupby("city")["revenue"].sum()

print("\n--- Sales Analysis ---")
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Average Order Value: ₹{avg_order_value:,.2f}")
print(f"Highest Value Order: ID {highest_order['order_id']} (₹{highest_order['revenue']:,.2f})")
print(f"Lowest Value Order: ID {lowest_order['order_id']} (₹{lowest_order['revenue']:,.2f})")

print("\nRevenue Product-Wise:")
print(product_revenue.to_string())

print("\nRevenue City-Wise:")
print(city_revenue.to_string())
```

```python
# 5. Save processed data
df.to_csv("sales_report.csv", index=False)
print("\nSaved processed data to 'sales_report.csv'.")
```
