```python
import os
import csv
import pandas as pd

# -------------------------------------------------------------
# Part 1: Product Classes (OOP & Inheritance)
# -------------------------------------------------------------
class Product:
    def __init__(self, product_id, name, brand, price, stock):
        self.product_id = product_id
        self.name = name
        self.brand = brand
        self.price = float(price)
        self.stock = int(stock)

    def display_product(self):
        return f"[{self.product_id}] {self.brand} {self.name} | Price: ₹{self.price} | Stock: {self.stock}"

    def update_stock(self, quantity):
        self.stock += quantity


class Electronics(Product):
    def __init__(self, product_id, name, brand, price, stock, warranty, model):
        super().__init__(product_id, name, brand, price, stock)
        self.category = "Electronics"
        self.warranty = warranty
        self.model = model

    def to_csv_row(self):
        return [self.product_id, self.name, self.brand, self.price, self.stock, self.category, self.warranty, self.model, "", "", "", ""]


class Clothing(Product):
    def __init__(self, product_id, name, brand, price, stock, size, material):
        super().__init__(product_id, name, brand, price, stock)
        self.category = "Clothing"
        self.size = size
        self.material = material

    def to_csv_row(self):
        return [self.product_id, self.name, self.brand, self.price, self.stock, self.category, "", "", self.size, self.material, "", ""]


class Books(Product):
    def __init__(self, product_id, name, brand, price, stock, author, publisher):
        super().__init__(product_id, name, brand, price, stock)
        self.category = "Books"
        self.author = author
        self.publisher = publisher

    def to_csv_row(self):
        return [self.product_id, self.name, self.brand, self.price, self.stock, self.category, "", "", "", "", self.author, self.publisher]
```

```python
# -------------------------------------------------------------
# Part 4 & 5: Pandas Analysis & Report Generation
# -------------------------------------------------------------
os.makedirs("output", exist_ok=True)

df_sales = pd.read_csv("sales.csv")
df_products = pd.read_csv("products.csv")

# Clean/Process sales dataframe
df_sales["total_amount"] = df_sales["quantity"] * df_sales["price"]

# Metrics
total_txns = len(df_sales)
total_qty = df_sales["quantity"].sum()
total_rev = df_sales["total_amount"].sum()
avg_tx_val = df_sales["total_amount"].mean()
highest_tx = df_sales.loc[df_sales["total_amount"].idxmax()]
lowest_tx = df_sales.loc[df_sales["total_amount"].idxmin()]

# Aggregations
product_sales_agg = df_sales.groupby(["product_id", "product_name"]).agg(
    units_sold=("quantity", "sum"),
    total_revenue=("total_amount", "sum")
).reset_index()

best_selling_prod = product_sales_agg.loc[product_sales_agg["units_sold"].idxmax()]["product_name"]
lowest_selling_prod = product_sales_agg.loc[product_sales_agg["units_sold"].idxmin()]["product_name"]

category_report = df_sales.groupby("category").agg(
    total_transactions=("transaction_id", "count"),
    units_sold=("quantity", "sum"),
    total_revenue=("total_amount", "sum")
).reset_index()

city_sales_report = df_sales.groupby("customer_city").agg(
    total_transactions=("transaction_id", "count"),
    total_revenue=("total_amount", "sum")
).reset_index()

payment_dist = df_sales["payment_method"].value_counts().reset_index()
payment_dist.columns = ["payment_method", "transaction_count"]

top_5_prods = product_sales_agg.sort_values(by="total_revenue", ascending=False).head(5)
bottom_5_prods = product_sales_agg.sort_values(by="total_revenue", ascending=True).head(5)

# Product catalog analysis
avg_prod_price = df_products["price"].mean()
max_prod_price = df_products["price"].max()
min_prod_price = df_products["price"].min()
low_stock_prods = df_products[df_products["stock"] < 10]
above_avg_price_prods = df_products[df_products["price"] > avg_prod_price]

# Highest category & city by revenue
highest_rev_cat = category_report.loc[category_report["total_revenue"].idxmax()]["category"]
highest_rev_city = city_sales_report.loc[city_sales_report["total_revenue"].idxmax()]["customer_city"]

# -------------------------------------------------------------
# Save Outputs to /output
# -------------------------------------------------------------
df_sales.to_csv("output/cleaned_sales.csv", index=False)
product_sales_agg.to_csv("output/product_report.csv", index=False)
city_sales_report.to_csv("output/city_sales_report.csv", index=False)
category_report.to_csv("output/category_report.csv", index=False)

summary_text = f"""==================================================
           RETAIL STORE SALES REPORT
==================================================
Total Transactions: {total_txns}
Total Units Sold: {total_qty}
Total Revenue: ₹{total_rev:,.2f}
Average Transaction Value: ₹{avg_tx_val:,.2f}

Best Selling Product (Units): {best_selling_prod}
Lowest Selling Product (Units): {lowest_selling_prod}
Highest Revenue Category: {highest_rev_cat}
Highest Revenue City: {highest_rev_city}

--------------------------------------------------
Product Catalog Insights:
- Average Catalog Price: ₹{avg_prod_price:,.2f}
- Max Catalog Price: ₹{max_prod_price:,.2f}
- Min Catalog Price: ₹{min_prod_price:,.2f}
- Items with Stock < 10: {len(low_stock_prods)}
- Items Priced Above Average: {len(above_avg_price_prods)}
=================================================="""

with open("output/sales_summary.txt", "w", encoding="utf-8") as f:
    f.write(summary_text)

print(summary_text)
print("\nGenerated files successfully under 'output/'.")
```
