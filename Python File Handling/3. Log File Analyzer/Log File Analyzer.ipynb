```python
import pandas as pd

# 1. Create and populate application.log with 30 entries
log_data = [
    "INFO - Application initialized\n",
    "INFO - Database connection pool created\n",
    "WARNING - Disk space running below 20%\n",
    "INFO - User 'admin' logged in\n",
    "ERROR - Database connection timeout\n",
    "INFO - Retrying database connection...\n",
    "INFO - Database connection re-established\n",
    "WARNING - High memory allocation detected (82%)\n",
    "ERROR - Payment gateway unreachable\n",
    "INFO - Incoming API request: /api/v1/orders\n",
    "WARNING - API response latency exceeded 500ms\n",
    "INFO - Cache hit ratio: 94%\n",
    "ERROR - Failed to send confirmation email to user #104\n",
    "INFO - Background worker job #42 finished\n",
    "INFO - User 'sudh' logged in\n",
    "WARNING - Rate limit approaching for IP 192.168.1.55\n",
    "ERROR - Invalid JWT token supplied\n",
    "INFO - User session terminated\n",
    "INFO - Health check status: OK\n",
    "WARNING - CPU temperature elevated\n",
    "ERROR - File write failed: permission denied on /var/log/app.lock\n",
    "INFO - Garbage collector executed\n",
    "INFO - Microservice 'auth' responding normally\n",
    "ERROR - Redis cluster node 2 unreachable\n",
    "WARNING - SSL certificate expires in 7 days\n",
    "INFO - Backup routine started\n",
    "INFO - Backup routine completed successfully\n",
    "ERROR - NullPointer exception encountered in worker thread #3\n",
    "WARNING - Unoptimized SQL query detected\n",
    "INFO - Server shutdown signal received\n",
]

with open("application.log", "w") as f:
    f.writelines(log_data)
```

```python
# 2. Analyze log entries using standard Python file handling
total_entries = 0
info_count = 0
warning_count = 0
error_count = 0

error_lines = []
warning_lines = []

with open("application.log", "r") as f:
    for line in f:
        clean_line = line.strip()
        if not clean_line:
            continue
        total_entries += 1
        
        if clean_line.startswith("INFO"):
            info_count += 1
        elif clean_line.startswith("WARNING"):
            warning_count += 1
            warning_lines.append(line)
        elif clean_line.startswith("ERROR"):
            error_count += 1
            error_lines.append(line)
```

```python
# 3. Write segregated entries to errors.txt and warnings.txt
with open("errors.txt", "w") as f:
    f.writelines(error_lines)

with open("warnings.txt", "w") as f:
    f.writelines(warning_lines)

print("--- Log Analysis Summary ---")
print(f"Total Log Entries: {total_entries}")
print(f"INFO Entries:      {info_count}")
print(f"WARNING Entries:   {warning_count}")
print(f"ERROR Entries:     {error_count}")
print("Generated 'errors.txt' and 'warnings.txt' successfully.")
```

```python
# 4. Bonus: Pandas Frequency Analysis
print("\n--- Pandas Frequency Analysis ---")
df = pd.read_csv(
    "application.log", 
    sep=" - ", 
    engine="python", 
    names=["Level", "Message"]
)

# Value counts & percentage frequency
level_counts = df["Level"].value_counts()
level_percentages = df["Level"].value_counts(normalize=True) * 100

summary_df = pd.DataFrame({
    "Count": level_counts,
    "Percentage (%)": level_percentages.round(2)
})
print(summary_df)
```
