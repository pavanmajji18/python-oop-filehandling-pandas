```python
# 2. Append new student records using 'a' mode
new_student = "111,Deepak,deepak@example.com,8765432109,AI Engineering,Noida\n"
with open("students.txt", "a") as f:
    f.write(new_student)
```

```python
# 3. Read complete records and read individual lines
print("--- Reading Individual Lines ---")
with open("students.txt", "r") as f:
    for line_num, line in enumerate(f, start=1):
        print(f"Record {line_num}: {line.strip()}")
```

```python
4. Count number of records
with open("students.txt", "r") as f:
    records = [line for line in f if line.strip()]
    total_records = len(records)
print(f"\nTotal Registered Students: {total_records}")
```

```python
# 5. Create a backup file (students_backup.txt)
with open("students.txt", "r") as src, open("students_backup.txt", "w") as dest:
    dest.write(src.read())
print("Backup created: students_backup.txt")
```

```python
# 6. Extract and create a file containing only names and courses
with open("students.txt", "r") as src, open("names_and_courses.txt", "w") as dest:
    dest.write("Name,Course\n")
    for line in src:
        parts = line.strip().split(",")
        if len(parts) >= 5:
            name, course = parts[1], parts[4]
            dest.write(f"{name},{course}\n")
print("Filtered list created: names_and_courses.txt")
```
