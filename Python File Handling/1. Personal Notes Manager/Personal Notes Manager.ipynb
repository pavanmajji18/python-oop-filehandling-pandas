```python
# 1. Create and write 10 initial notes using write mode ('w') with standard open/close
file = open("notes.txt", "w")
initial_notes = [
    "Note 1: Review Python file handling basics.\n",
    "Note 2: Practice open(), read(), write(), and close().\n",
    "Note 3: Understand difference between read(), readline(), and readlines().\n",
    "Note 4: Remember to use context managers (with open).\n",
    "Note 5: File modes: 'r' is read-only, 'w' truncates/overwrites.\n",
    "Note 6: File mode 'a' appends to the end without truncating.\n",
    "Note 7: Always handle newlines explicitly when writing lines.\n",
    "Note 8: Relative vs absolute file paths.\n",
    "Note 9: Encoding defaults usually to UTF-8.\n",
    "Note 10: Clean up file descriptors to prevent leaks.\n",
]
file.writelines(initial_notes)
file.close()
```

```python
# 2. Reading specific number of characters and entire file using 'r' mode
file = open("notes.txt", "r")
print("--- First 20 Characters ---")
print(file.read(20))

print("\n--- Remaining File Content ---")
print(file.read())
file.close()
```

```python
# 3. Demonstrating readline() vs readlines()
file = open("notes.txt", "r")
print("--- Using readline() for first two lines ---")
print("Line 1:", file.readline().strip())
print("Line 2:", file.readline().strip())

print("\n--- Using readlines() for the rest ---")
remaining_lines = file.readlines()
for idx, line in enumerate(remaining_lines, start=3):
    print(f"Line {idx}: {line.strip()}")
file.close()
```

```python
# 4. Appending additional notes using append mode ('a') with `with open(...)`
new_notes = [
    "Note 11: Added using append ('a') mode.\n",
    "Note 12: Context managers automatically close files upon exit.\n",
]
with open("notes.txt", "a") as f:
    f.writelines(new_notes)
```

```python
# 5. Reopening to display the fully updated contents
print("\n--- Final Updated File Content ---")
with open("notes.txt", "r") as f:
    print(f.read())
```
