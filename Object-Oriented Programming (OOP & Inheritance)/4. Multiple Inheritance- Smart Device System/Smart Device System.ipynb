```python
class Camera:
    def take_photo(self):
        return "Capturing high-resolution 48MP photo..."

    def record_video(self):
        return "Recording 4K 60fps video..."


class MusicPlayer:
    def play_music(self, track):
        return f"Playing track: '{track}'"

    def stop_music(self):
        return "Music playback stopped."


class GPS:
    def current_location(self):
        return "Coordinates: 17.3850° N, 78.4867° E"

    def navigate(self, destination):
        return f"Navigating to {destination} using optimal route..."


class SmartPhone(Camera, MusicPlayer, GPS):
    def __init__(self, brand, model, price, storage):
        self.brand = brand
        self.model = model
        self.price = price
        self.storage = storage

    def phone_specs(self):
        return f"{self.brand} {self.model} | {self.storage}GB | ₹{self.price:,.2f}"
```

```python
# 3 SmartPhone Objects
phones = [
    SmartPhone("Apple", "iPhone 15 Pro", 134900, 256),
    SmartPhone("Samsung", "Galaxy S24 Ultra", 129999, 512),
    SmartPhone("Google", "Pixel 8 Pro", 106999, 128),
]

for phone in phones:
    print(phone.phone_specs())
    print(" ", phone.take_photo())
    print(" ", phone.play_music("Midnight City"))
    print(" ", phone.navigate("Airport Terminal 2"))
    print()
```
