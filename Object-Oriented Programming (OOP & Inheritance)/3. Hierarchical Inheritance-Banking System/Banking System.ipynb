```python
class BankAccount:
    def __init__(self, account_number, holder_name, balance=0.0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ₹{amount:,.2f} into Account {self.account_number}")

    def display_balance(self):
        print(f"Balance for {self.account_number} ({self.holder_name}): ₹{self.balance:,.2f}")

    def account_details(self):
        return f"Acc No: {self.account_number} | Holder: {self.holder_name} | Balance: ₹{self.balance:,.2f}"


class SavingsAccount(BankAccount):
    def __init__(self, account_number, holder_name, balance, interest_rate):
        super().__init__(account_number, holder_name, balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        interest = self.balance * (self.interest_rate / 100)
        self.balance += interest
        print(f"Interest added: ₹{interest:,.2f} at {self.interest_rate}%")


class CurrentAccount(BankAccount):
    def __init__(self, account_number, holder_name, balance, overdraft_limit):
        super().__init__(account_number, holder_name, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        if amount <= self.balance + self.overdraft_limit:
            self.balance -= amount
            print(f"Withdrawn ₹{amount:,.2f}. Current balance: ₹{self.balance:,.2f}")
        else:
            print("Withdrawal exceeds overdraft limit!")


class SalaryAccount(BankAccount):
    def __init__(self, account_number, holder_name, balance, employer, monthly_salary):
        super().__init__(account_number, holder_name, balance)
        self.employer = employer
        self.monthly_salary = monthly_salary

    def credit_salary(self):
        self.deposit(self.monthly_salary)
        print(f"Salary credited from {self.employer}")
```

```python
# Demonstration
sa = SavingsAccount("SB-1001", "Rohan Sharma", 50000, interest_rate=4.5)
ca = CurrentAccount("CA-2001", "Apex Enterprises", 150000, overdraft_limit=50000)
sal_acc = SalaryAccount("SAL-3001", "Anita Verma", 10000, employer="Infosys", monthly_salary=85000)

sa.display_balance()
sa.apply_interest()

ca.withdraw(180000)

sal_acc.credit_salary()
sal_acc.display_balance()
```
