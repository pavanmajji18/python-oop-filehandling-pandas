```python
class Employee:
    def __init__(self, employee_id, name, salary, department):
        self.employee_id = employee_id
        self.name = name
        self.salary = salary
        self.department = department

    def display_details(self):
        print(f"ID: {self.employee_id} | Name: {self.name} | Dept: {self.department} | Salary: ${self.salary:,.2f}")

    def calculate_salary(self):
        return self.salary


class Developer(Employee):
    def __init__(self, employee_id, name, salary, department, programming_language):
        super().__init__(employee_id, name, salary, department)
        self.programming_language = programming_language

    def display_details(self):
        super().display_details()
        print(f"  └─ Core Language: {self.programming_language}")


class Manager(Employee):
    def __init__(self, employee_id, name, salary, department, team_size):
        super().__init__(employee_id, name, salary, department)
        self.team_size = team_size

    def display_details(self):
        super().display_details()
        print(f"  └─ Team Size: {self.team_size} members")


class HR(Employee):
    def __init__(self, employee_id, name, salary, department, region):
        super().__init__(employee_id, name, salary, department)
        self.region = region

    def display_details(self):
        super().display_details()
        print(f"  └─ Region: {self.region}")
```

```python
# Object Creation (2 objects per child class)
dev1 = Developer("D101", "Alice Chen", 95000, "Engineering", "Python")
dev2 = Developer("D102", "Bob Smith", 92000, "Engineering", "Go")

mgr1 = Manager("M201", "Carol Taylor", 125000, "Operations", 12)
mgr2 = Manager("M202", "David Ross", 130000, "Engineering", 8)

hr1 = HR("H301", "Elena Cruz", 78000, "Human Resources", "North America")
hr2 = HR("H302", "Farhan Malik", 81000, "Human Resources", "APAC")
```

```python
# Demonstration
for emp in [dev1, dev2, mgr1, mgr2, hr1, hr2]:
    emp.display_details()
```
