```python
class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

    def get_personal_info(self):
        return f"Person: {self.name}, Age: {self.age}, City: {self.city}"


class Employee(Person):
    def __init__(self, name, age, city, employee_id, salary, company):
        super().__init__(name, age, city)
        self.employee_id = employee_id
        self.salary = salary
        self.company = company

    def get_employment_info(self):
        return f"Company: {self.company} | ID: {self.employee_id} | Salary: ${self.salary:,.2f}"


class Developer(Employee):
    def __init__(self, name, age, city, employee_id, salary, company, language, framework, experience):
        super().__init__(name, age, city, employee_id, salary, company)
        self.language = language
        self.framework = framework
        self.experience = experience

    def get_developer_profile(self):
        return (
            f"{self.get_personal_info()}\n"
            f"{self.get_employment_info()}\n"
            f"Tech Stack: {self.language} ({self.framework}) | Experience: {self.experience} yrs\n"
        )
```

```python
# 5 Developer Objects
devs = [
    Developer("Aarav", 26, "Bengaluru", "E101", 85000, "TechCorp", "Python", "FastAPI", 3),
    Developer("Priya", 29, "Hyderabad", "E102", 110000, "CloudScale", "JavaScript", "React", 6),
    Developer("Liam", 24, "Austin", "E103", 78000, "DevStudio", "Rust", "Actix", 2),
    Developer("Mei", 32, "Singapore", "E104", 135000, "FinTechX", "Java", "Spring Boot", 9),
    Developer("Carlos", 28, "Berlin", "E105", 98000, "DataFlow", "Go", "Gin", 5),
]
```

```python
# Demonstrating access across all 3 levels
for dev in devs:
    print(dev.get_developer_profile())
```
