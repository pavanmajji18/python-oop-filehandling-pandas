```python
class Payment:
    def process_payment(self, amount):
        print(f"Base Payment: Processing transaction of ₹{amount:,.2f}")


class CreditCardPayment(Payment):
    def __init__(self, card_number):
        self.card_number = card_number[-4:]

    def process_payment(self, amount):
        print(f"Processing ₹{amount:,.2f} using Credit Card ending in **{self.card_number} (2% surcharge applied).")


class UPIPayment(Payment):
    def __init__(self, upi_id):
        self.upi_id = upi_id

    def process_payment(self, amount):
        print(f"Processing ₹{amount:,.2f} using UPI (VPA: {self.upi_id}) via instant settlement.")


class NetBankingPayment(Payment):
    def __init__(self, bank_name):
        self.bank_name = bank_name

    def process_payment(self, amount):
        print(f"Processing ₹{amount:,.2f} using NetBanking via {self.bank_name} secure gateway.")


class WalletPayment(Payment):
    def __init__(self, wallet_name):
        self.wallet_name = wallet_name

    def process_payment(self, amount):
        print(f"Processing ₹{amount:,.2f} using {self.wallet_name} Wallet balance.")
```

```python
# Demonstration of Polymorphism via Method Overriding
transactions = [
    (CreditCardPayment("4111222233334567"), 12500),
    (UPIPayment("user@oksbi"), 5000),
    (NetBankingPayment("HDFC Bank"), 45000),
    (WalletPayment("Paytm"), 850),
]

for payment_method, amt in transactions:
    payment_method.process_payment(amt)
```
