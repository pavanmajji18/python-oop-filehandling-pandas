```python
import numpy as np
import pandas as pd

# 1. Generate & Save Dataset (50 Products)
np.random.seed(42)
categories = [
    'Laptop',
    'Mobile',
    'Headphones',
    'Keyboard',
    'Monitor',
    'Tablet',
    'Accessories',
]
brands = ['Apple', 'Samsung', 'Dell', 'Sony', 'Logitech', 'Boat', 'Asus', 'HP']

data_p2 = {
    'product_id': [f'PROD_{i:03d}' for i in range(1, 51)],
    'product_name': [f'Product {i}' for i in range(1, 51)],
    'category': np.random.choice(categories, size=50),
    'brand': np.random.choice(brands, size=50),
    'price': np.random.randint(1500, 120000, size=50),
    'rating': np.round(np.random.uniform(2.5, 5.0, size=50), 1),
    'stock': np.random.randint(1, 50, size=50),
    'discount': np.random.choice(
        [5, 10, 15, 20, 25], size=50
    ),  # discount in %
}
pd.DataFrame(data_p2).to_csv('products.csv', index=False)
```

```python
# 2. Pandas Operations
df_products = pd.read_csv('products.csv')

# Dataset Inspection & Unique Counts
print('Total Products:', len(df_products))
print('Unique Categories:', df_products['category'].unique())
print('Unique Brands:', df_products['brand'].unique())
print('\nCategory-wise Count:\n', df_products['category'].value_counts())
```

```python
# Price & Rating Queries
print(f"\nAverage Price: ₹{df_products['price'].mean():.2f}")
print(
    'Highest-Priced Product:\n',
    df_products.loc[df_products['price'].idxmax()],
)
print(
    'Lowest-Priced Product:\n', df_products.loc[df_products['price'].idxmin()]
)

high_price_products = df_products[df_products['price'] > 50000]
high_rating_products = df_products[df_products['rating'] > 4.0]
low_stock_products = df_products[df_products['stock'] < 10]
```

```python
# Sorting
sorted_price = df_products.sort_values(by='price', ascending=False)
sorted_rating = df_products.sort_values(by='rating', ascending=False)
```

```python
# Calculated Columns
df_products['discount_amount'] = (
    df_products['price'] * (df_products['discount'] / 100)
).round(2)
df_products['final_price'] = (
    df_products['price'] - df_products['discount_amount']
).round(2)
```

```python
# Aggregations
avg_price_by_cat = df_products.groupby('category')['price'].mean()
avg_rating_by_brand = df_products.groupby('brand')['rating'].mean()
max_price_by_cat = df_products.groupby('category')['price'].max()

print('\nAverage Price by Category:\n', avg_price_by_cat)
print('\nAverage Rating by Brand:\n', avg_rating_by_brand)
print('\nMax Price by Category:\n', max_price_by_cat)
```

```python
# Export Final Data
df_products.to_csv('output_ecommerce_analysis.csv', index=False)
```
