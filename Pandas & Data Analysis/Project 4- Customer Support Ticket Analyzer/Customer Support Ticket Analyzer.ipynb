```python
import numpy as np
import pandas as pd

# 1. Generate & Save Dataset (50 Tickets)
np.random.seed(42)
categories = ['Payment', 'Technical', 'Login', 'Course', 'Refund', 'Account']
priorities = ['Low', 'Medium', 'High', 'Critical']
statuses = ['Open', 'In Progress', 'Resolved', 'Closed']
agents = ['Rahul', 'Pooja', 'Vikram', 'Ananya', 'Suresh']

data_p4 = {
    'ticket_id': [f'TICK_{i:04d}' for i in range(1, 51)],
    'customer_name': [f'Customer_{i}' for i in range(1, 51)],
    'category': np.random.choice(categories, size=50),
    'priority': np.random.choice(priorities, size=50, p=[0.3, 0.3, 0.25, 0.15]),
    'status': np.random.choice(statuses, size=50),
    'assigned_agent': np.random.choice(agents, size=50),
    'resolution_time': np.random.randint(15, 360, size=50),  # in minutes
}
pd.DataFrame(data_p4).to_csv('support_tickets.csv', index=False)
```

```python
# 2. Pandas Operations & Analysis
df_tickets = pd.read_csv('support_tickets.csv')

print('Total Tickets:', len(df_tickets))
print('\nTickets by Category:\n', df_tickets['category'].value_counts())
print('\nTickets by Priority:\n', df_tickets['priority'].value_counts())
print('\nTickets by Status:\n', df_tickets['status'].value_counts())

# Filter Queries
open_tickets = df_tickets[df_tickets['status'] == 'Open']
critical_tickets = df_tickets[df_tickets['priority'] == 'Critical']
resolved_tickets = df_tickets[df_tickets['status'] == 'Resolved']

# Resolution Time Analysis
print(
    f"\nOverall Average Resolution Time: {df_tickets['resolution_time'].mean():.2f} mins"
)

avg_res_by_cat = df_tickets.groupby('category')['resolution_time'].mean()
print('\nAverage Resolution Time per Category (mins):\n', avg_res_by_cat)

agent_ticket_counts = df_tickets.groupby('assigned_agent')['ticket_id'].count()
print('\nTickets Handled per Agent:\n', agent_ticket_counts)

# Sorting & Extreme Values
sorted_tickets = df_tickets.sort_values(
    by='resolution_time', ascending=False
)
max_res_time_ticket = df_tickets.loc[df_tickets['resolution_time'].idxmax()]
print('\nTicket with Longest Resolution Time:\n', max_res_time_ticket)
```

```python
# Export Summary Report
summary_report = (
    df_tickets.groupby(['category', 'priority'])
    .agg(
        total_tickets=('ticket_id', 'count'),
        avg_resolution_mins=('resolution_time', 'mean'),
        max_resolution_mins=('resolution_time', 'max'),
    )
    .reset_index()
)

summary_report.to_csv('support_ticket_summary_report.csv', index=False)
print(
    '\nSummary report successfully exported to support_ticket_summary_report.csv'
)
```
