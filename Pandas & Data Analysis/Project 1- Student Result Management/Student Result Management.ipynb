```python
import os
import numpy as np
import pandas as pd

# 1. Generate & Save Dataset (25 Students)
np.random.seed(42)
names = [
    'Aarav',
    'Aditi',
    'Ananya',
    'Bhavya',
    'Chetan',
    'Deepak',
    'Divya',
    'Gaurav',
    'Ishaan',
    'Kavya',
    'Manish',
    'Neha',
    'Nitin',
    'Pooja',
    'Pranav',
    'Priya',
    'Rahul',
    'Riya',
    'Rohan',
    'Sakshi',
    'Siddharth',
    'Sneha',
    'Tanvi',
    'Varun',
    'Yash',
]
cities = [
    'Mumbai',
    'Delhi',
    'Bengaluru',
    'Hyderabad',
    'Pune',
    'Chennai',
    'Kolkata',
]

data_p1 = {
    'student_id': range(101, 126),
    'name': names,
    'age': np.random.randint(19, 24, size=25),
    'city': np.random.choice(cities, size=25),
    'python_marks': np.random.randint(50, 100, size=25),
    'sql_marks': np.random.randint(50, 100, size=25),
    'pandas_marks': np.random.randint(50, 100, size=25),
    'attendance': np.random.randint(60, 100, size=25),
}
df_p1 = pd.DataFrame(data_p1)
df_p1.to_csv('students.csv', index=False)
```

```python
# 2. Pandas Operations
df = pd.read_csv('students.csv')

print('--- First 5 Records --- \n', df.head())
print('\n--- Last 5 Records --- \n', df.tail())
print('\nDataset Shape:', df.shape)
print('Column Names:', df.columns.tolist())
print('\nData Types:\n', df.dtypes)
print('\n--- Info ---')
df.info()
print('\n--- Describe --- \n', df.describe())

print(f"\nAverage Python Marks: {df['python_marks'].mean():.2f}")
print(f"Average SQL Marks: {df['sql_marks'].mean():.2f}")
print(f"Average Pandas Marks: {df['pandas_marks'].mean():.2f}")
```

```python
# Highest and Lowest marks across subject columns
numeric_marks = ['python_marks', 'sql_marks', 'pandas_marks']
print(f"Maximum Mark Overall: {df[numeric_marks].max().max()}")
print(f"Minimum Mark Overall: {df[numeric_marks].min().min()}")
```

```python
# Sorting
sorted_by_python = df.sort_values(by='python_marks', ascending=False)
sorted_by_attendance = df.sort_values(by='attendance', ascending=False)
```

```python
# Filtering & Column Selection
python_gt_80 = df[df['python_marks'] > 80]
attendance_gt_75 = df[df['attendance'] > 75]
selected_cols = df[['name', 'python_marks', 'pandas_marks']]
```

```python
# Add Computed Columns
df['total_marks'] = (
    df['python_marks'] + df['sql_marks'] + df['pandas_marks']
)
df['average_marks'] = (df['total_marks'] / 3).round(2)
```

```python
# Save Processed Dataset
os.makedirs('output', exist_ok=True)
df.to_csv('output/student_results.csv', index=False)
print('\nSuccessfully saved to output/student_results.csv')
```
